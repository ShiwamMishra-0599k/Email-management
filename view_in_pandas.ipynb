```python
import chromadb
import pandas as pd

# Connect to ChromaDB
client = chromadb.PersistentClient(path="./chroma_store")
collection = client.get_collection(name="email_data")

# Fetch all records
results = collection.get(include=['metadatas', 'documents'])


# Prepare list of dicts for DataFrame
data = []
for i in range(len(results['ids'])):
    row = {
        "email_id": results['ids'][i],
        "sender": results['metadatas'][i].get("sender"),
        "subject": results['metadatas'][i].get("subject"),
        "summary": results['metadatas'][i].get("summary"),
        "urgency": results['metadatas'][i].get("urgency"),
        "intent": results['metadatas'][i].get("intent"),
        "topic": results['metadatas'][i].get("topic"),
        "auto_reply": results['metadatas'][i].get("auto_reply"),
        "ai_reply": results['metadatas'][i].get("ai_reply"),
        "sentiment": results['metadatas'][i].get("sentiment"),
        "body": results['documents'][i][:200] 
    }
    data.append(row)

# Create DataFrame
df = pd.DataFrame(data)

# Display nicely
print(df.to_string(index=False))
```
